```python
# ============================================
# 3. Conexión a MongoDB Atlas con SSL
# ============================================
# Reemplaza <usuario>, <password> y <cluster> con tus credenciales de Atlas
from pymongo import MongoClient
import certifi

uri = "mongodb+srv://<usuario>:<password>@cluster0.kapoqn2.mongodb.net/?appName=<cluster>"
try:
    client = MongoClient(uri, tls=True, tlsCAFile=certifi.where())
    info = client.server_info()  # Esto fuerza el handshake SSL
    print("✅ Handshake SSL exitoso")
    print("Versión del servidor:", info["version"])
except Exception as e:
    print("❌ Error en el handshake SSL:", e)
```

```python

from pymongo.mongo_client import MongoClient
from pymongo.server_api import ServerApi

uri = "mongodb+srv://proyecto:123@cluster0.n4znbfu.mongodb.net/?appName=Cluster0"

# Create a new client and connect to the server
client = MongoClient(uri, server_api=ServerApi('1'))

# Send a ping to confirm a successful connection
try:
    client.admin.command('ping')
    print("Pinged your deployment. You successfully connected to MongoDB!")
except Exception as e:
    print(e)
```
